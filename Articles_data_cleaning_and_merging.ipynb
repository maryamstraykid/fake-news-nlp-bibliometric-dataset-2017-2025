```python
from habanero import Crossref
import pandas as pd

cr = Crossref()

# Load CSV with article titles
df = pd.read_csv("analyse bibliol ieee.csv")

citation_counts = []

for title in df["Title"]:
    try:
        result = cr.works(query=title)
        if "message" in result and "items" in result["message"] and len(result["message"]["items"]) > 0:
            citations = result["message"]["items"][0].get("is-referenced-by-count", 0)
            citation_counts.append(citations)
            print(f"Citations for {title}: {citations}")
        else:
            citation_counts.append(None)
    except Exception as e:
        citation_counts.append(None)
        print(f"Error fetching {title}: {e}")

# Add citation data to DataFrame
df["Citations"] = citation_counts

# Save results
df.to_csv("articles_with_citations.csv", index=False)

print("Citation data saved successfully!")
```

```python
# Remplacer NaN dans "Author Keywords" par "Index Keywords"
df_scopus["Author Keywords"].fillna(df_scopus["Index Keywords"], inplace=True)
```

```python
# Drop the "Index Keywords" column
df_scopus.drop(columns=["Index Keywords"], inplace=True)
```

```python
df_goscho.drop(columns=["Key"], inplace=True)
```

```python
df_iee.drop(columns=["Publication Title"], inplace=True)
```

```python
# Standardize column names
column_mapping = {
    "Authors": "Authors",
    "Author": "Authors",
    "Title": "Title",
    "Year": "Year",
    "Sourcetitle": "Conference Name",
    "Source title": "Conference Name",
    "Publication Title": "Conference Name",
    "Conference Name": "Conference Name",
    "Publication Year": "Year",
    "Cited by": "Citations",
    "Citations": "Citations",
    "DOI": "DOI",
    "Affiliations": "Affiliations",
    "Abstract": "Abstract",
    "Abstract Note": "Abstract",
    "Author Keywords": "Keywords",
    "Automatic Tags": "Keywords",
    "Library Catalog": "Source",
    "Source": "Source",
    "Item Type": "Document Type"
    }
```

```python
# Rename columns in each dataset
df_scopus = df_scopus.rename(columns=column_mapping)
df_iee = df_iee.rename(columns=column_mapping)
df_goscho = df_goscho.rename(columns=column_mapping)
```

```python
# Define the desired column order
desired_order = ['Document Type', 'Year', 'Authors', 'Title',
                 'Conference Name', 'DOI', 'Abstract', 'Source',
                 'Citations', 'Keywords']

# Reorder the DataFrame columns
df_iee = df_iee[desired_order]
```

```python
# Define the desired column order
desired_order = ['Document Type', 'Year', 'Authors', 'Title',
                 'Conference Name', 'DOI', 'Abstract', 'Source',
                 'Citations', 'Keywords','Affiliations']

# Reorder the DataFrame columns
df_scopus = df_scopus[desired_order]
```

```python
# Merge the datasets using an outer join
df_merged = pd.concat([df_iee, df_goscho, df_scopus], ignore_index=True, sort=False)
```

```python
# Remove rows where "Author" is missing
df_merged = df_merged.dropna(subset=["Authors"])
```

```python
df_merged = df_merged.dropna(subset=["Conference Name"])
```

```python
df_merged = df_merged.dropna(subset=["Abstract"])
```

```python
df_merged = df_merged.dropna(subset=["DOI"])
```

```python
df_merged['Year'] = df_merged['Year'].astype('Int64')
```
